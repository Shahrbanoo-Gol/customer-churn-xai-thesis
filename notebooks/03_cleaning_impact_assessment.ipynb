# Step 3 — Cleaning impact assessment

This notebook presents the main findings produced by `assess_cleaning_impact.py`. The analysis is deliberately non-destructive: it evaluates temporary scenarios and does not apply final cleaning, define churn, create predictive features, or fit a model.

```python
from pathlib import Path
import pandas as pd

ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
TABLES = ROOT / 'outputs' / 'tables'
def load_table(name):
    return pd.read_csv(TABLES / name)
missing = load_table('31_missing_customer_impact.csv')
duplicates = load_table('32_duplicate_impact_summary.csv')
cancellations = load_table('34_customer_cancellation_profile.csv')
roles = load_table('38_transaction_role_summary.csv')
decisions = load_table('39_cleaning_impact_decision_table.csv')
```

## Main findings

```python
display(missing.query("section == 'summary'")[['category', 'metric', 'value', 'percentage', 'notes']])
display(duplicates)
display(cancellations.query("section == 'summary'"))
display(roles.query("row_type == 'primary_role'"))
display(decisions[['issue', 'recommended_treatment', 'confidence']])
```
